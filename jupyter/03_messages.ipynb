```python
import os

from IPython.terminal.shortcuts.auto_suggest import accept
from dotenv import load_dotenv
load_dotenv(override=True)
```

```python
from langchain.agents import create_agent
from langchain.tools import tool
from langchain_core.messages import SystemMessage,HumanMessage,AIMessage
from langchain_openai import ChatOpenAI
import os
```

```python
model = ChatOpenAI(
    model = "deepseek-ai/DeepSeek-V4-Flash",
    api_key=os.getenv("GUIJI_API_KEY"),
    base_url=os.getenv("GUIJI_BASE_URL"),
)
@tool
def get_weather(location:str)->str:
    ## 工具函数必须有文档参数
    """Get the weather in a location."""
    return f"The weather in {location} is milk."
agent = create_agent(
    model=model,
    tools=[get_weather]
)
```

```python
response = agent.invoke({
    "messages": [
        SystemMessage(content="请使用工具来过去天气"),
        HumanMessage(content="你好我是牛子"),
        AIMessage(content="你好，我是一个助手。很高兴认识你"),
        HumanMessage(content="今天北京的天气如何")
    ]
})
print(response)
```

```python
res = agent.stream({
    "messages": [
        HumanMessage(content="我是谁？你是谁？")
    ]
},stream_mode="messages")

for token,metadata in res:
    print(token.content,end="",flush=True)
```

```python
from langchain_openai import ChatOpenAI
import os
from dotenv import load_dotenv
load_dotenv(override=True)
model = ChatOpenAI(
    model='Qwen/Qwen3.6-35B-A3B',
    base_url=os.getenv("GUIJI_BASE_URL"),
    api_key=os.getenv("GUIJI_API_KEY"),
)
agent = create_agent(model=model)
```

```python
message = HumanMessage([
    {"type":"text","text":"描述下面这张图片"},
    {"type":"image","url":"https://help-static-aliyun-doc.aliyuncs.com/file-manage-files/zh-CN/20241022/emyrja/dog_and_girl.jpeg"}
])
```

```python
stream = agent.stream(
    {
        "messages":[message]
    },stream_mode = "messages"
)
for chunk,metadata in stream:
    print(chunk.content,end = "",flush=True)
```

```python
# 创建一个上传组件，用模拟图片上传
from ipywidgets import FileUpload
from IPython.display import display
uploader = FileUpload(accept = '*',multiple = False)
display(uploader)
```

```python
print(uploader.value)
```

```python
# 读取图片
import base64

uploaded_file = uploader.value[0]
content_mv = uploaded_file["content"]
img_bytes = bytes(content_mv)
img_b64 = base64.b64encode(img_bytes).decode("utf-8")
```

```python
# 组织多模态消息
multimode_question = HumanMessage(
    content=[
        {
            "type":"image",
            "base64":img_b64,
            "mime_type":"image/jpeg",
        },{
           "type":"text","text":"给我讲讲图片中的内容"
        }
    ]
)
```

```python
for chunk,metadata in agent.stream({"messages":[multimode_question]},stream_mode="messages"):
    print(chunk.content,end="",flush=True)
```
