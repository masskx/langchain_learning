```python
from langgraph import graph
from langgraph.constants import START,END
from langgraph.graph import StateGraph
# 模拟搭建一个搜索系统
from typing import TypedDict


# 定义一个状态
class MyState(TypedDict): # 通过状态管理数据
    query:str
    rag_result:str
    web_result:str
    final_answer:str

# 定义节点
def rag_search_node(state:MyState):
    # 从状态中获取数据
    query = state['query']
    # 模拟构建一些搜索结果
    rag_result = f"这是基于{query}的搜索结果"
    # 将处理好的值返回到相应的状态的值
    return {
        "rag_result":rag_result
    }

def web_search_node(state:MyState):
    # 从状态中获取数据
    query = state['query']
    # 模拟构建一个web搜索结果
    web_result = f"这是基于{query}的搜索结果"
    # 将处理后的结果返回到状态中
    return {
        "web_result":web_result
    }

def fianl_answer_node(state:MyState):
    # 从状态中获取数据
    query = state['query']
    rag_result = state['rag_result']
    web_result = state['web_result']
    # 构建模拟结果
    final_answer = f"基于{query}的网络搜索结果是{web_result},rag结果是{rag_result}"

    # 将处理好的结果进行返回到状态
    return {
        "final_answer":final_answer
    }

# 创建构造对象，开始构造图结构

builder = StateGraph(state_schema=MyState)
# 将构造好的就节点加入构造器
builder.add_node(rag_search_node)
builder.add_node(web_search_node)
builder.add_node(fianl_answer_node)
# 将注册进来的节点通过边实现链接
builder.add_edge(START, "rag_search_node")
builder.add_edge("rag_search_node", "web_search_node")
builder.add_edge("web_search_node", "fianl_answer_node")
builder.add_edge("fianl_answer_node", END)
# 将构造器对象进行编译，得到可运行的图对象
graph = builder.compile()
# 调用图对象，通过invoke传入初始值
response = graph.invoke({
    "query":"如何使用langgraph"
})
print(response)

```
