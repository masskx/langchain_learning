```python
from dotenv import load_dotenv
from langchain_core.messages import HumanMessage

load_dotenv(override=True)
```

```python
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI
import os
from langchain.messages import HumanMessage
model = ChatOpenAI(
    model="Qwen/Qwen3.6-35B-A3B",
    base_url=os.getenv("GUIJI_BASE_URL"),
    api_key=os.getenv("GUIJI_API_KEY")
)

agent = create_agent(
    model=model
)

response = agent.stream({
    "messages":[HumanMessage(content="你是谁？")]
},stream_mode="messages")

for chunk,metadata in response:
    print(chunk.content,end="",flush=True)
```

```python
# 尝试系统提示词
from langchain.agents import create_agent
from langchain.messages import HumanMessage

agent = create_agent(
    model = model,
    system_prompt="你以高悦的口吻向我说话，至于高悦是谁，她是我对象"
)

response = agent.stream({
    "messages":[HumanMessage("宝宝我们去吃啥？")]
},stream_mode="messages")
# 调用
for chunk,metadata in response:
    print(chunk.content,end = "",flush=True)
```

```python
# 设定角色和指令
system_prompt = """
你是我的女朋友，你叫高悦，咱们俩感情很好，你对我也非常好。
"""
# 创建智能体
agent = create_agent(model = model, system_prompt=system_prompt)
response = agent.stream({"messages":[HumanMessage(content="宝宝别生气了，我下次会注意的（委屈）我不是故意的么")]},stream_mode="messages")

for chunk,metadata in response:
    print(chunk.content,end="",flush=True)
```

```python
system_prompt = """
# 角色
- 我女朋友
# 名字
-高悦
# 指令
- 最近把你惹生气了，你不理我，也不说话，我很难受
- 我尝试哄你，但是这次很严重，几乎决裂
"""
# 创建智能体
agent = create_agent(model=model, system_prompt=system_prompt)
response = agent.stream({"messages":[HumanMessage(content="姐姐，我真的知道错了，是我不对")]},stream_mode="messages")
for chunk,metadata in response:
    print(chunk.content,end="",flush=True)
```
