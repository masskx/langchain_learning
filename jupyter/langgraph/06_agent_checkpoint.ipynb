```python

# Agent检查点机制
from dotenv import load_dotenv

load_dotenv(override=True)
from common import init_llm
from langchain_core.tools import tool
from langchain.agents import create_agent
llm = init_llm.deepseek_llm

@tool
def get_weather(city: str):
    """
    查天气的工具
    """
    return f"{city}的天气是大暴雨天"
from langgraph.checkpoint.memory import InMemorySaver
checkpointer = InMemorySaver()

agent = create_agent(
    model=llm,
    tools=[get_weather],
    checkpointer=checkpointer
)
config = {
    "configurable":{
        "thread_id":"1"
    }
}
#用户第一次调用
res1 = agent.invoke({
    "messages":"陕西的天气怎么样？"
},config=config)
print("第一次调用",res1['messages'][-1])
print("+=================================")
#用户第二次调用
res2 = agent.invoke({
    "messages":"出门建议怎么穿衣服？适合出门嘛"
},config=config)
print("第二次调用",res2['messages'][-1])

```
